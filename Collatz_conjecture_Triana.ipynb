# Collatz Conjecture Sequence Length Optimization

This project computes the longest Collatz sequence for all starting integers $n \le 10^6$ using dynamic programming with memoization in Python.

## Problem Description

The Collatz sequence is defined for positive integers as follows:

$$f(n) =  \begin{cases}  \frac{n}{2} & \text{if } n \text{ is even} \\ 3n + 1 & \text{if } n \text{ is odd} \end{cases}$$

The Collatz Conjecture states that repeatedly applying $f(n)$ will eventually reach $1$ for all positive integers $n$.

The objective of this project is to find the integer $n \le 10^6$ that produces the maximum sequence length $I(n)$ to reach $1$.

## Algorithm and Implementation

The implementation utilizes a recursive function with **memoization** via a global dictionary to store previously computed sequence lengths, reducing redundant subproblem evaluations from $O(N \cdot L)$ down to nearly $O(N)$ amortized time.

```python
from numba import jit, njit
import time
jitted_collatz = jit(nopython=True)(collatz)
dictionary = {1: 0}
def collatz(n):
    if n in dictionary:
        return dictionary[n]
    if n % 2 == 0:
        result = 1 + collatz(n // 2)
    else:
        result = 1 + collatz(3 * n + 1)
    dictionary[n] = result
    return result

t0 = time.time()
for n in range(1, 10**6 + 1):
    collatz(n)
t1 = time.time() - t0

max_n = max(range(1, 10**6+1), key=lambda n: dictionary[n])
max_I = dictionary[max_n]

print("Max I(n) = ",max_I,  " at  n = ", max_n)
print("Time elapsed: " ,t1, " seconds")
```
