# High-Performance Python Numerical Computation Optimization

This project benchmarks and compares execution performance across different Python numerical acceleration toolkits (`NumPy`, `NumExpr`, and `Numba` JIT compilation) applied to sequence generation, array operations, and numerical integration.

## Problem Description

Standard CPython loops introduce overhead when processing large-scale numerical arrays or iterative algorithms. This repository evaluates performance optimizations across three common mathematical tasks:

1. **Element-wise Vector Operations:** Evaluating trigonometric expressions on $10^7$ element arrays.
2. **Iterative Sequence Generation:** Computing Fibonacci numbers up to $N = 92$[cite: 2].
3. **Numerical Integration:** Approximating definite integrals using the Trapezoidal Rule[cite: 2].

## Benchmarked Approaches and Technologies

* **NumPy:** Baseline vectorized array operations[cite: 2].
* **NumExpr:** Fast numerical expression evaluator that avoids intermediate memory allocations and utilizes multi-threading[cite: 2].
* **Numba JIT (`@jit`, `@njit`):** Just-In-Time compilation translating Python/NumPy code directly into optimized machine code via LLVM[cite: 2].

@jit
def trapezoidal_optimized(f, a, b, N):
    h = (b - a) / N
    x = np.linspace(a, b, N + 1)
    y = f(x)
    total_sum = np.sum(y[:-1] + y[1:])
    return total_sum * h / 2

````python
def trapezoidal_regular(f, a, b, N):
    h = (b-a)/N
    x = np.linspace(a,b,N+1)
    y = f(x)
    integral = 0.0
    for i in range (N):
        integral += h/2*(y[i]+y[i+1])
    return integral
````

````python
@jit
def trapezoidal_optimized(f, a, b, N):
    h = (b-a)/N
    x = np.linspace(a,b,N+1)
    y = f(x)
    sum = np.sum(y[:-1]+y[1:])
    integral = sum*h/2

    return integral
````
